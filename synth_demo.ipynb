```python
import numpy as np
import json
import glob
import sys

sys.path.append("./synthetic/")
from synthetic.utils.synth import place_single_object, place_multiple_objects
from synthetic.utils.viz import visualize_bboxes

with open("synthetic/class.json", "r") as f:
    CLASS = json.load(f)
    f.close()
    
class_names = {v:k for k, v in CLASS.items()}
class_names
```

### Single Object

```python
segment_files = np.sort(glob.glob("data/klsg_prep/masked/ship/*"))
bg_files = glob.glob("data/klsg_dataset/mine/*")

segment_file = np.random.choice(segment_files)
bg_file = np.random.choice(bg_files)
composed, bbox = place_single_object(bg_file, segment_file, "obj-001", save=False)

visualize_bboxes(composed, [bbox], color=(253, 141, 60), class_names=class_names)
```

### Multiple object

```python
composed, bboxes, class_ids = place_multiple_objects(bg_files, segment_files, "mixed-001", save=False)

visualize_bboxes(composed, bboxes, color=(253, 141, 60), class_names=class_names)
```
